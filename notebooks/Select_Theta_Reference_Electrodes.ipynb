## Make a bunch of plots to help us pick reference electrodes

```python
restriction = "nwb_file_name LIKE 'IM-1478_20220726%'"

# fetch_good_units picks the right sorting pipeline and curation for each session:
# Berke lab IM-* sessions are spikesorting v1 (curation 1 = not-whitened), Frank lab
# sessions are v0 (curation 2 = after manual curation). Pass curation_id= to override.

# One theta entry per (session, run epoch). Frank lab sessions have 4-5 run epochs, so we
# iterate over entries rather than over nwb file names
theta_entries = (HexMazeThetaV1() & restriction).fetch(
    "nwb_file_name", "target_interval_list_name", as_dict=True)
print(f"{len(theta_entries)} theta entries to plot")

for entry in theta_entries:
    nwb_file_name = entry["nwb_file_name"]
    target_interval = entry["target_interval_list_name"]
    session_id = f"{nwb_file_name.replace('_.nwb', '')}_{target_interval}"

    theta_key = {"nwb_file_name": nwb_file_name, "target_interval_list_name": target_interval}
    epoch = (sgc.TaskEpoch & {"nwb_file_name": nwb_file_name,
                              "interval_list_name": target_interval}).fetch1("epoch")
    decode_key = {"nwb_file_name": nwb_file_name, "epoch": epoch}
    print(f"\n{session_id}: interval {target_interval} = epoch {epoch}")
    
    # Save figures as PDFs, one folder per session
    FIG_DIR = Path("figures") / session_id
    FIG_DIR.mkdir(parents=True, exist_ok=True)

    def save_fig(fig, name):
        """Save a figure as a PDF into FIG_DIR."""
        path = FIG_DIR / f"{name}.pdf"
        fig.savefig(path, bbox_inches="tight")
        print(f"saved {path}")

    # Well-isolated units for this session, from whichever pipeline its lab uses
    good = fetch_good_units(nwb_file_name)

    # Electrodes carrying a good unit (the units' peak channels). Empty for Frank lab v0
    # sessions, which have no peak channel -- those plots just show no unit markers.
    unit_electrodes = electrodes_with_units(good)

    print(f"{len(good)} accepted units on {len(unit_electrodes)} electrodes")
    print(unit_electrodes)

    # Time window to plot (seconds from the start of the recording)
    start_time, duration = 100.0, 1.5

    # Electrodes that carry a good unit (from the units cell above)
    unit_elecs = set(unit_electrodes)

    # Open the raw NWB and pull a short window of all channels
    with pynwb.NWBHDF5IO(Nwbfile().get_abs_path(nwb_file_name), "r", load_namespaces=True) as io:
        nwbf = io.read()
        eseries = get_raw_electrical_series(nwbf)   # "ElectricalSeries" (Berke) or "e-series" (Frank)
        ts = np.asarray(eseries.timestamps)

        # Convert the time window to sample indices
        i0, i1 = int(np.searchsorted(ts, start_time)), int(np.searchsorted(ts, start_time + duration))
        t = ts[i0:i1] - ts[i0]                                  # time within the window (s)
        data = np.asarray(eseries.data[i0:i1, :], dtype=float)  # raw traces (n_time, n_chan) in µV

        # Electrode-table row for each data column (gives geometry + bad-channel tags)
        elec = nwbf.electrodes.to_dataframe().iloc[list(eseries.electrodes.data)]

    # Remember each electrode's data column, and keep its electrode_id (== peak_channel id)
    elec["data_col"] = range(len(elec))
    elec["electrode_id"] = elec.index

    # Order channels left-to-right by shank, then top-to-bottom within shank (rel_y descending)
    elec = elec.sort_values(["probe_shank", "rel_y"], ascending=[True, False]).reset_index(drop=True)

    # Set the trace scale from good channels only
    scale = 4 * np.nanstd(data[:, elec.loc[elec["bad_channel"] == 0, "data_col"].astype(int)])

    # Plot each channel as a stacked trace, colored by shank (bad channels faded)
    n = len(elec)
    fig, ax = plt.subplots(figsize=(18, 0.15 * n + 2))
    rows = []        # (y position, electrode name, electrode id/number, shank color, has-unit) per channel
    unit_ys = []     # y positions of channels with a good unit, for the stars
    for k, row in elec.iterrows():
        y = n - 1 - k # stack top-to-bottom (first row on top)

        # Alternate trace color by shank; fade bad channels to gray
        base = "black" if row["probe_shank"] % 2 == 0 else "#cc0000"
        color = base if row["bad_channel"] == 0 else ("#aaaaaa" if base == "black" else "#dd9999")
        ax.plot(t, data[:, int(row["data_col"])] / scale + y, color=color, lw=0.4, rasterized=True)

        # Note channels that carry a good unit so we can mark them after the loop
        has_unit = row["electrode_id"] in unit_elecs
        if has_unit:
            unit_ys.append(y)
        rows.append((y, row["electrode_name"], int(row["electrode_id"]), base, has_unit))

    # Star the channels that carry a good unit (white fill + black edge to match the theta phase plot)
    ax.scatter([t[0]] * len(unit_ys), unit_ys, marker="*", s=90, c="white",
            edgecolors="black", linewidths=0.6, zorder=5, clip_on=False, label="has good unit")

    # Label each trace with its electrode number AND name (number first so you can read off channels),
    # colored by shank; bold if it has a unit
    ax.set_yticks([r[0] for r in rows])
    ax.set_yticklabels([f"{eid:>4}  {name}" for (_, name, eid, _, _) in rows], fontsize=5, fontfamily="monospace")
    for lab, (_, _, _, base, has_unit) in zip(ax.get_yticklabels(), rows):
        lab.set_color(base)
        lab.set_fontweight("bold" if has_unit else "normal")

    ax.set_ylim(-0.7, n - 0.3)
    ax.set_xlabel("Time (s)")
    ax.legend(loc="upper left", fontsize=8)
    ax.set_title(f"{nwb_file_name} raw traces {start_time}–{start_time + duration}s "
                f"({len(unit_elecs)} channels with good units)")
    plt.tight_layout()
    save_fig(fig, f"{session_id}_raw_traces")
    plt.show()

    SPEED_THRESH = 5.0       # cm/s: only keep running times (for decode-distance measures only)
    SPATIAL_COV_MAX = 200.0  # only keep confident deocdes (for decode-distance measures only)
    #HIGH_THETA_PCT = 50.0   # optionally keep only high theta-power times (based on the reference electrode)
    SUBSAMPLE = 50

    # Get theta phase and theta power for each electrode (df columns "electrode N", indexed by time)
    phase_df = (HexMazeThetaV1() & theta_key).fetch1_theta_phase()
    power_df = (HexMazeThetaV1() & theta_key).fetch1_theta_power()
    # Subsample theta power and phase so they aren't huge
    theta_phase_sub, theta_power_sub = phase_df.iloc[::SUBSAMPLE], power_df.iloc[::SUBSAMPLE]

    # Get probe geometry (rel_x, rel_y) and bad_channel tag for each electrode
    elec_table = fetch_electrode_geometry(nwb_file_name)
    is_bad = elec_table["is_bad"]

    # Use electrodes that are not "bad channel" and in the theta table
    # I think sometimes electrodes that are ecog / ground / ref screws are sometimes not marked bad
    # based on impedance, but aren't included in theta tables. TODO confirm this
    theta_elecs = {c.split()[-1] for c in phase_df.columns}      # electrode names that have a theta column
    good_elecs = set(elec_table.index[~is_bad.values])           # electrodes flagged good (bad_channel == False)
    missing_good = sorted(good_elecs - theta_elecs, key=int)     # good but missing from theta
    if missing_good:
        print(f"WARNING: {len(missing_good)} good electrode(s) missing from the theta table, "
            f"skipping them: {missing_good}")
    usable_elecs = [n for n in elec_table.index if n in good_elecs and n in theta_elecs]  # good AND has a theta column

    # Find the good channel with the highest median theta power so we can use that as our reference electrode
    median_theta_power = power_df.median()
    median_theta_power.index = median_theta_power.index.str.split().str[-1] # Convert index from "electrode N" to "N"
    good_power = median_theta_power[usable_elecs]   # good electrodes that actually have a theta column
    ref_name = good_power.idxmax()
    ref_col = f"electrode {ref_name}"

    # Fetch and align speed, decode distance, and spatial cov to subsampled theta timestamps
    decode_df = (HexMazeDecodedPosition & decode_key).fetch1_dataframe()
    speed = decode_df["speed"].reindex(theta_phase_sub.index, method="nearest")
    decode_dist = decode_df["decode_distance"].reindex(theta_phase_sub.index, method="nearest")
    spatial_coverage = decode_df["spatial_cov"].reindex(theta_phase_sub.index, method="nearest")

    # 1: Get theta phase lag of each electrode relative to the reference electrode
    reference_theta_phase = theta_phase_sub[ref_col].values
    theta_phase_offsets = {name: np.degrees(np.angle(np.mean(np.exp(
                1j * (theta_phase_sub[f"electrode {name}"].values - reference_theta_phase)))))
            for name in usable_elecs}

    # 2: Get correlation between each electrode's theta power envelope and the reference electrode's
    theta_power_corr = theta_power_sub.corrwith(theta_power_sub[ref_col])
    theta_power_corr.index = theta_power_corr.index.str.split().str[-1]

    # 3: Get pearson correlation of theta power with speed for all electrodes
    power_speed_corr = theta_power_sub.corrwith(speed)
    power_speed_corr.index = power_speed_corr.index.str.split().str[-1] # Convert index from "electrode N" to "N"

    # Filter by speed and spatial_cov for decode distance based measures
    # (Uncomment the theta_power_thresh lines below to also filter for high theta times)
    # theta_power_thresh = np.nanpercentile(theta_power_sub[ref_col], HIGH_THETA_PCT)
    times_to_keep = (
        (speed.values >= SPEED_THRESH)
        # & (theta_power_sub[ref_col].values >= theta_power_thresh)
        & (spatial_coverage.values < SPATIAL_COV_MAX)
        & decode_dist.notna().values
    )

    kept_phase = theta_phase_sub.values[times_to_keep]      # (n_samples, n_electrodes) theta phase in radians
    kept_decode_dist = decode_dist.values[times_to_keep]    # (n_samples,) signed decode distance from rat
    cos_phase, sin_phase = np.cos(kept_phase), np.sin(kept_phase)

    # Vectorized pearson corr of `vector` against each column of `matrix` (one value per electrode)
    def columnwise_pearson(matrix, vector):
        matrix_centered = matrix - matrix.mean(0)
        vector_centered = vector - vector.mean()
        return (matrix_centered * vector_centered[:, None]).sum(0) / np.sqrt(
            (matrix_centered ** 2).sum(0) * (vector_centered ** 2).sum())

    corr_dist_cos = columnwise_pearson(cos_phase, kept_decode_dist)   # corr(decode distance, cos phase) per electrode
    corr_dist_sin = columnwise_pearson(sin_phase, kept_decode_dist)   # corr(decode distance, sin phase) per electrode
    # corr(cos, sin) per electrode
    cos_centered, sin_centered = cos_phase - cos_phase.mean(0), sin_phase - sin_phase.mean(0)
    corr_cos_sin = (cos_centered * sin_centered).sum(0) / np.sqrt((cos_centered ** 2).sum(0) * (sin_centered ** 2).sum(0))

    # 4: Get correlation of decode distance with theta phase for each electrode 
    # Theta phase is circular so use Mardia's circular-linear correlation (which handles this by combining 
    # decode distance's correlation with cos(phase) and with sin(phase) into one number in [0, 1])
    # (0 = no relationship to phase, 1 = phase fully predicts decode distance)
    # Says how strong the theta phase / decode distance relationship is, but not where peak is (see 5 for that)
    circ_lin_corr = np.sqrt(
        (corr_dist_cos ** 2 + corr_dist_sin ** 2 - 2 * corr_dist_cos * corr_dist_sin * corr_cos_sin)
        / (1 - corr_cos_sin ** 2))
    dist_phase_corr = pd.Series(circ_lin_corr, index=[c.split()[-1] for c in theta_phase_sub.columns])

    # 5: Get theta phase at which decode distance is most positive (decoded position furthest ahead of the rat)
    # First-harmonic fit decode_distance ~ a*cos(phase) + b*sin(phase) peaks at phase = atan2(b, a); 
    # the correlation slopes corr_dist_cos, corr_dist_sin give the same peak angle
    pref_phase_deg = np.degrees(np.arctan2(corr_dist_sin, corr_dist_cos))   # [-180, 180]
    dist_pref_phase = pd.Series(pref_phase_deg, index=[c.split()[-1] for c in theta_phase_sub.columns])

    print(f"Reference electrode: {ref_name}   (n = {times_to_keep.sum()} samples)")
    print(f"Preferred theta phase at reference electrode: {dist_pref_phase[ref_name]:.0f} deg")

    # Add every metric to the electrode geometry table for plotting (NaN out bad channels)
    electrode_geometry = elec_table.copy()
    electrode_geometry["phase_offset_deg"] = pd.Series(theta_phase_offsets)
    electrode_geometry["theta_power"] = median_theta_power
    electrode_geometry["power_corr"] = theta_power_corr
    electrode_geometry["power_speed_corr"] = power_speed_corr
    electrode_geometry["dist_phase_corr"] = dist_phase_corr
    electrode_geometry["dist_pref_phase_deg"] = dist_pref_phase
    metric_cols = ["phase_offset_deg", "theta_power", "power_corr",
                "power_speed_corr", "dist_phase_corr", "dist_pref_phase_deg"]
    electrode_geometry.loc[is_bad, metric_cols] = np.nan
    has_unit = electrode_geometry.index.astype(int).isin(unit_elecs)

    # Get box size to plot based on the min spacing between electrodes rel_x and rel_y
    def min_spacing(v):
        d = np.diff(np.unique(v)); d = d[d > 0]
        return d.min() if len(d) else 1.0
    box_width, box_height = min_spacing(electrode_geometry["rel_x"].values), min_spacing(electrode_geometry["rel_y"].values)

    # Draw probe geometry with boxes for each electrode colored by 'values'
    # Add stars to the electrodes with good units and outline the reference electrode in yellow
    def geometry_boxes(ax, values, cmap_name, label, clim=None, n_categories=None, highlight=None):
        # Discrete colormap when coloring by category (e.g. layer labels), else continuous
        if n_categories is not None:
            cmap = plt.colormaps[cmap_name].resampled(n_categories).copy()
            norm = BoundaryNorm(np.arange(-0.5, n_categories + 0.5), n_categories)
        else:
            cmap, norm = plt.colormaps[cmap_name].copy(), None
        cmap.set_bad("white")
        boxes = [Rectangle((x - box_width/2, y - box_height/2), box_width, box_height) 
                for x, y in zip(electrode_geometry["rel_x"], electrode_geometry["rel_y"])]
        patch_collection = PatchCollection(boxes, cmap=cmap, norm=norm)
        patch_collection.set_array(np.ma.masked_invalid(values))
        if clim and n_categories is None: patch_collection.set_clim(*clim)
        ax.add_collection(patch_collection)
        # Mark electrodes that have good units with a star
        ax.scatter(electrode_geometry.loc[has_unit, "rel_x"], electrode_geometry.loc[has_unit, "rel_y"], marker="*", s=70,
                c="white", edgecolors="black", linewidths=0.6, zorder=5, label="Has good unit(s)")
        # Highlight reference electrode in yellow
        ax.add_patch(Rectangle((electrode_geometry.loc[ref_name, "rel_x"] - box_width/2, 
                                electrode_geometry.loc[ref_name, "rel_y"] - box_height/2),
                    box_width, box_height, fill=False, edgecolor="yellow", linewidth=2, label="Reference electrode"))
        # Outline a custom set of electrodes (e.g. the chosen reference set) in red
        if highlight is not None:
            for i, e in enumerate(highlight):
                ax.add_patch(Rectangle((electrode_geometry.loc[e, "rel_x"] - box_width/2,
                                        electrode_geometry.loc[e, "rel_y"] - box_height/2),
                            box_width, box_height, fill=False, edgecolor="red", linewidth=2,
                            zorder=6, label="Selected reference" if i == 0 else None))
        ax.set_xlim(electrode_geometry["rel_x"].min() - box_width, electrode_geometry["rel_x"].max() + box_width)
        ax.set_ylim(electrode_geometry["rel_y"].min() - box_height, electrode_geometry["rel_y"].max() + box_height)
        # Shape the axes box based on min/max x/y coords (otherwise the colorbar is huge and weird)
        xspan = (electrode_geometry["rel_x"].max() + box_width) - (electrode_geometry["rel_x"].min() - box_width)
        yspan = (electrode_geometry["rel_y"].max() + box_height) - (electrode_geometry["rel_y"].min() - box_height)
        ax.set_box_aspect(yspan / xspan)
        ax.set_ylabel("rel_y (µm)")
        plt.colorbar(patch_collection, ax=ax, label=label, fraction=0.02, pad=0.01,
                    ticks=range(n_categories) if n_categories is not None else None)

    # Get colormap limits
    speed_corr_limit = np.nanmax(np.abs(electrode_geometry["power_speed_corr"].values)) # symmetric scale around 0
    circ_lin_corr_limit = np.nanmax(electrode_geometry["dist_phase_corr"].values)

    # Big plot with all per-electrode measures. Pass highlight=[electrode names] to outline a
    # chosen set (e.g. the reference electrodes) on every panel.
    def plot_geometry_panels(highlight=None, save_name=None):
        fig, axes = plt.subplots(6, 1, figsize=(16, 15))
        geometry_boxes(axes[0], np.log10(electrode_geometry["theta_power"].values), "viridis",
                    "Theta power (log10)", highlight=highlight)
        geometry_boxes(axes[1], electrode_geometry["phase_offset_deg"].values, "twilight",
                    "Phase offset (deg)", clim=(-180, 180), highlight=highlight)
        geometry_boxes(axes[2], electrode_geometry["power_corr"].values, "magma",
                    "Corr with ref power", clim=(0, 1), highlight=highlight)
        geometry_boxes(axes[3], electrode_geometry["power_speed_corr"].values, "RdBu_r",
                    "Corr(theta power, speed)", clim=(-speed_corr_limit, speed_corr_limit), highlight=highlight)
        geometry_boxes(axes[4], electrode_geometry["dist_phase_corr"].values, "magma",
                    "Circ-lin corr(decode dist, theta phase)", clim=(0, circ_lin_corr_limit), highlight=highlight)
        geometry_boxes(axes[5], electrode_geometry["dist_pref_phase_deg"].values, "twilight",
                    "Preferred phase (deg)", clim=(-180, 180), highlight=highlight)
        axes[0].set_title("Median absolute theta power")
        axes[1].set_title("Theta phase offset from reference electrode")
        axes[2].set_title("Theta power correlation with reference electrode")
        axes[3].set_title("Theta power vs. speed correlation (per electrode)")
        axes[4].set_title(f"Decode distance vs. theta phase correlation (speed > {SPEED_THRESH}cm/s, spatial_cov < {SPATIAL_COV_MAX})")
        axes[5].set_title(f"Theta phase where decode distance is most positive (speed > {SPEED_THRESH}cm/s, spatial_cov < {SPATIAL_COV_MAX})")
        axes[0].legend(loc="upper left", fontsize=8)
        axes[5].set_xlabel("rel_x (µm)")
        fig.suptitle(nwb_file_name, fontsize=13)
        plt.tight_layout()
        if save_name:
            save_fig(fig, save_name)
        plt.show()


    plot_geometry_panels(save_name=f"{session_id}_theta_metrics_probe_geometry")

    # Labeled 6-panel probe map for manually picking a reference set: same panels as
    # plot_geometry_panels above, but with each GOOD electrode's number printed on its box so you
    # can read off the numbers of the channels you want on every metric at once. Bad channels are
    # left blank. Reuses geometry_boxes / electrode_geometry / is_bad / box_width / box_height and
    # the colormap limits (speed_corr_limit / circ_lin_corr_limit) from above.
    import matplotlib.patheffects as pe

    # Numbers sit slightly up-and-right of the box center so they don't land on top of the unit
    # stars (which geometry_boxes draws at the box center). Offsets are a fraction of the box size.
    NUM_DX, NUM_DY = 0.22, 0.22

    def label_electrode_numbers(ax):
        """Write each good electrode's number on its box, offset off-center so it clears the star.
        White text + black outline stays readable on any box color; bad channels are skipped."""
        for name, row in electrode_geometry.iterrows():
            if is_bad.get(name, True) or pd.isna(row["rel_x"]):
                continue
            ax.text(row["rel_x"] + NUM_DX * box_width, row["rel_y"] + NUM_DY * box_height, name,
                    ha="center", va="center", fontsize=4, color="white", fontweight="bold",
                    path_effects=[pe.withStroke(linewidth=1.0, foreground="black")], zorder=7)

    # Build the same 6 panels as plot_geometry_panels (theta power is shown on a log scale)
    fig, axes = plt.subplots(6, 1, figsize=(16, 18))
    geometry_boxes(axes[0], np.log10(electrode_geometry["theta_power"].values), "viridis",
                "Theta power (log10)")
    geometry_boxes(axes[1], electrode_geometry["phase_offset_deg"].values, "twilight",
                "Phase offset (deg)", clim=(-180, 180))
    geometry_boxes(axes[2], electrode_geometry["power_corr"].values, "magma",
                "Corr with ref power", clim=(0, 1))
    geometry_boxes(axes[3], electrode_geometry["power_speed_corr"].values, "RdBu_r",
                "Corr(theta power, speed)", clim=(-speed_corr_limit, speed_corr_limit))
    geometry_boxes(axes[4], electrode_geometry["dist_phase_corr"].values, "magma",
                "Circ-lin corr(decode dist, theta phase)", clim=(0, circ_lin_corr_limit))
    geometry_boxes(axes[5], electrode_geometry["dist_pref_phase_deg"].values, "twilight",
                "Preferred phase (deg)", clim=(-180, 180))

    # Write the electrode numbers on every panel
    for ax in axes:
        label_electrode_numbers(ax)

    axes[0].set_title("Median absolute theta power")
    axes[1].set_title("Theta phase offset from reference electrode")
    axes[2].set_title("Theta power correlation with reference electrode")
    axes[3].set_title("Theta power vs. speed correlation (per electrode)")
    axes[4].set_title(f"Decode distance vs. theta phase correlation (speed > {SPEED_THRESH}cm/s, spatial_cov < {SPATIAL_COV_MAX})")
    axes[5].set_title(f"Theta phase where decode distance is most positive (speed > {SPEED_THRESH}cm/s, spatial_cov < {SPATIAL_COV_MAX})")
    axes[0].legend(loc="upper left", fontsize=8)
    axes[5].set_xlabel("rel_x (µm)")
    fig.suptitle(f"{nwb_file_name} — good electrode numbers on every metric panel "
                "(for manually picking a reference set)", fontsize=13)
    plt.tight_layout()
    save_fig(fig, f"{session_id}_theta_metrics_with_numbers")
    plt.show()

    # Group electrodes into layers based on theta metrics:
    # Electrodes in the same layer should have consistent theta phase (offset from the reference) and theta power
    # Reuses electrode_geometry / is_bad / has_unit / geometry_boxes from the cell above.
    LAYER_COUNTS = [3, 4, 5, 6]   # candidate numbers of layers: compared and plotted for each
    W_PHASE = 2.0              # weight of theta phase relative to theta power
    W_POWER = 1.0              # weight of theta power

    # Good channels with the measures we cluster on
    good_mask = ~is_bad & electrode_geometry[["phase_offset_deg", "theta_power"]].notna().all(axis=1)
    good_electrodes = electrode_geometry.loc[good_mask]
    phase_rad = np.radians(good_electrodes["phase_offset_deg"].values)
    log_power = np.log10(good_electrodes["theta_power"].values) # power spans orders of magnitude

    # Build the feature matrix. Circular theta phase becomes [cos, sin] so the 360 deg wrap is
    # handled. Standardize so phase and power are comparable, then up-weight phase.
    features = StandardScaler().fit_transform(
        np.column_stack([np.cos(phase_rad), np.sin(phase_rad), log_power])
    ) * np.array([W_PHASE, W_PHASE, W_POWER])

    # Compare numbers of layers on the metrics that actually discriminate (phase_R saturates
    # near 1 at any k because theta phase is a smooth depth gradient, so it can't pick k).
    # Higher silhouette = better separated; lower Davies-Bouldin = better separated.
    scan_rows = []
    for k in LAYER_COUNTS:
        trial = KMeans(n_clusters=k, n_init=10, random_state=0).fit_predict(features)
        scan_rows.append({
            "k": k,
            "mean_silhouette": round(silhouette_score(features, trial), 3),
            "davies_bouldin": round(davies_bouldin_score(features, trial), 3),
        })
    print("Ccomparison between number of layers (higher silhouette / lower Davies-Bouldin = better separated):")
    print("(Note we don't expect these to be super informative because we have a smooth gradient vs real clusters)")
    display(pd.DataFrame(scan_rows).set_index("k"))


    def circular_stats(angles_rad):
        """Mean angle (deg) and concentration R of a set of angles.
        R is the mean resultant length: 1 = all identical, 0 = uniformly spread."""
        resultant = np.mean(np.exp(1j * angles_rad))
        R = np.abs(resultant)
        circular_std_deg = np.degrees(np.sqrt(-2 * np.log(R))) if R > 0 else np.inf
        return np.degrees(np.angle(resultant)), R, circular_std_deg


    def layers_for_k(n_layers):
        """Cluster electrodes into n_layers. Label numbers are arbitrary (KMeans order)."""
        return KMeans(n_clusters=n_layers, n_init=10, random_state=0).fit_predict(features)


    # Show the grouping for each requested number of layers
    for n_layers in LAYER_COUNTS:
        layer = layers_for_k(n_layers)
        sample_silhouette = silhouette_samples(features, layer)  # per-electrode separation score
        print(f"\nSplit into {n_layers} layers (overall silhouette = {silhouette_score(features, layer):.3f}):")

        # Per-layer consistency: tight phase (high R / low std) and small power spread = consistent
        rows = []
        for lyr in range(n_layers):
            in_layer = layer == lyr
            phase_mean, phase_R, phase_std = circular_stats(phase_rad[in_layer])
            rows.append({
                "layer": lyr,
                "n": int(in_layer.sum()),
                "phase_mean_deg": round(phase_mean, 1),
                "phase_R": round(phase_R, 3),              # 1 = identical phase, 0 = spread out
                "phase_std_deg": round(phase_std, 1),      # circular spread of phase within the layer
                "log_power_mean": round(log_power[in_layer].mean(), 2),
                "log_power_std": round(log_power[in_layer].std(), 2),
                "mean_silhouette": round(sample_silhouette[in_layer].mean(), 3),
            })
        display(pd.DataFrame(rows).set_index("layer"))

        # Full-length label array for the geometry plot (NaN for bad / excluded channels)
        layer_full = pd.Series(np.nan, index=electrode_geometry.index)
        layer_full.loc[good_electrodes.index] = layer

        # Geometry map + feature-space scatter for this grouping
        fig, axes = plt.subplots(2, 1, figsize=(16, 6))
        geometry_boxes(axes[0], layer_full.values, "tab10", "layer", n_categories=n_layers)
        axes[0].set_title(f"{session_id} electrodes split into {n_layers} layers")
        axes[0].set_xlabel("rel_x (µm)")

        discrete_cmap = plt.colormaps["tab10"].resampled(n_layers)
        layer_norm = BoundaryNorm(np.arange(-0.5, n_layers + 0.5), n_layers)
        scatter = axes[1].scatter(good_electrodes["phase_offset_deg"], log_power, c=layer,
                                cmap=discrete_cmap, norm=layer_norm,
                                edgecolors="k", linewidths=0.3)
        axes[1].set_xlabel("Theta phase offset from reference (deg)")
        axes[1].set_ylabel("log10 theta power")
        axes[1].set_title(f"{n_layers} layers in power/phase feature space (one point per electrode)")
        plt.colorbar(scatter, ax=axes[1], label="layer", ticks=range(n_layers))
        plt.tight_layout()
        save_fig(fig, f"{session_id}_layers_k{n_layers}")
        plt.show()
```

## Evaluate reference electrode pick

```python
import numpy as np
import spyglass.common as sgc
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from pathlib import Path

from spyglass_hexmaze.hex_maze_decoding import (
    HexMazeDecodedPosition, HexMazeThetaV1, HexMazeThetaReference,
)
from spyglass_hexmaze.spikesorting_helpers import (
    fetch_good_units, electrodes_with_units, fetch_electrode_geometry,
)

nwb_file_name = "Lily20251217_.nwb"
REFERENCE_NAME = "theta_reference_1"

# Pick run epoch. 
# Frank lab sessions have 4-5 run epochs ("01_r1", "03_r2", "05_r3", ...)
# Berke lab sessions have a single "00_r1".
# This has to be pinned because theta and decode are keyed differently: 
# HexMazeThetaV1 uses target_interval_list_name (because of LFP), 
# but HexMazeDecodedPosition keys on the epoch number. 
# Restricting by nwb_file_name alone can silently land them on different epochs, in which case
# no spike falls inside the theta window and every downstream count comes out zero :(
# Look the epoch number up from TaskEpoch so both restrictions describe the same epoch.
TARGET_INTERVAL = "05_r3"

epoch = (sgc.TaskEpoch & {"nwb_file_name": nwb_file_name,
                          "interval_list_name": TARGET_INTERVAL}).fetch1("epoch")
theta_key = {"nwb_file_name": nwb_file_name, "target_interval_list_name": TARGET_INTERVAL}
decode_key = {"nwb_file_name": nwb_file_name, "epoch": epoch}
print(f"{nwb_file_name}  interval {TARGET_INTERVAL} = epoch {epoch}")

# Fetch reference electrode set for this nwb
ref = HexMazeThetaReference & {**theta_key, "reference_name": REFERENCE_NAME}
reference_electrodes = list(ref.fetch1("electrode_ids"))
print(f"loaded '{REFERENCE_NAME}': {len(reference_electrodes)} electrodes {reference_electrodes}")

reference_electrodes = list(reference_electrodes)
ref_set = {str(e) for e in reference_electrodes}   # electrode names are strings in the geometry index
session_id = nwb_file_name.replace("_.nwb", "")

# Save figures as PDFs, one folder per session (same convention as the rest of the notebook)
FIG_DIR = Path("figures") / session_id
FIG_DIR.mkdir(parents=True, exist_ok=True)
def save_fig(fig, name):
    """Save a figure as a PDF into FIG_DIR."""
    path = FIG_DIR / f"{name}.pdf"
    fig.savefig(path, bbox_inches="tight")
    print(f"saved {path}")

# Probe geometry (rel_x, rel_y) + bad-channel tag for each electrode, from the tables
electrode_geometry = fetch_electrode_geometry(nwb_file_name)
is_bad = electrode_geometry["is_bad"]

# Box size for the probe map: smallest nonzero spacing between electrode rel_x / rel_y
def min_spacing(v):
    d = np.diff(np.unique(v)); d = d[d > 0]
    return d.min() if len(d) else 1.0
box_width = min_spacing(electrode_geometry["rel_x"].values)
box_height = min_spacing(electrode_geometry["rel_y"].values)

# Good units (with spike_times) for this session, from whichever pipeline this lab uses
good_units = fetch_good_units(nwb_file_name)
unit_electrodes = electrodes_with_units(good_units)

# Build the theta reference: average the analytic signal across the picked electrodes.
# fetch1_reference_analytic() does exactly this for the saved electrode set, and reads only
# those electrodes' columns off disk -- the whole analytic signal table is ~21 GB on a
# 384-channel session, so pulling all of it to use one electrode costs ~10 minutes.
ref_complex = ref.fetch1_reference_analytic()             # complex reference theta, time-indexed
z_mean = ref_complex.to_numpy()                           # averaged complex theta

# Per-channel signals, for the cross-channel coherence check further down. Only the reference
# electrodes are read, so this stays cheap even on high channel count sessions.
analytic = (HexMazeThetaV1 & theta_key).fetch_analytic_signal(reference_electrodes)
z = np.column_stack([(analytic[f"electrode {e}_real"] + 1j * analytic[f"electrode {e}_imag"]).values
                     for e in reference_electrodes])
theta_phase_ref = pd.Series(np.angle(z_mean), index=analytic.index, name="theta_phase")
theta_power_ref = pd.Series(np.abs(z_mean) ** 2, index=analytic.index, name="theta_power")

# Phase-averaging quality: each channel's mean phase + cross-channel coherence (1 = in phase)
channel_phase = np.angle(z)
channel_mean_phase = np.angle(np.mean(np.exp(1j * channel_phase), axis=0))
coherence = np.abs(np.mean(np.exp(1j * channel_phase), axis=1)).mean()

# Align decode distance and keep running, confident samples
decode_df = (HexMazeDecodedPosition & decode_key).fetch1_dataframe()
dist = decode_df["decode_distance"].reindex(theta_phase_ref.index, method="nearest")
speed = decode_df["speed"].reindex(theta_phase_ref.index, method="nearest")
cov = decode_df["spatial_cov"].reindex(theta_phase_ref.index, method="nearest")
keep = (speed.values >= 5.0) & (cov.values < 200.0) & dist.notna().values
phase_deg = np.degrees(theta_phase_ref.values[keep]) % 360   # 0 = theta peak, 180 = trough
dist_kept = dist.values[keep]

def circular_linear_corr(phase_rad, linear):
    rxc = np.corrcoef(linear, np.cos(phase_rad))[0, 1]
    rxs = np.corrcoef(linear, np.sin(phase_rad))[0, 1]
    rcs = np.corrcoef(np.cos(phase_rad), np.sin(phase_rad))[0, 1]
    return np.sqrt((rxc ** 2 + rxs ** 2 - 2 * rxc * rxs * rcs) / (1 - rcs ** 2))

clc = circular_linear_corr(np.radians(phase_deg), dist_kept)

# Binned decode distance vs theta phase
edges = np.linspace(0, 360, 19)
centers = 0.5 * (edges[:-1] + edges[1:])
bin_of = np.clip(np.digitize(phase_deg, edges) - 1, 0, len(centers) - 1)
mean_dist = np.array([dist_kept[bin_of == b].mean() for b in range(len(centers))])
sem_dist = np.array([pd.Series(dist_kept[bin_of == b]).sem() for b in range(len(centers))])
peak_phase = centers[np.nanargmax(mean_dist)]


# SPIKE -> THETA PHASE: where do this session's cells fire relative to the reference theta phase?
# CA1 pyramidal cells fire near the theta TROUGH of pyramidal-layer theta, so the pooled spike phase 
# tells us where our reference phase's 0 (= waveform peak) sits relative to that (ideally we would see ~180)
SPIKE_SPEED_THRESH = 5.0   # cm/s: only phase-lock spikes that occur during running theta
speed_series = decode_df["speed"]

def spike_phases_deg(spike_times):
    """Reference theta phase (deg, 0=peak/180=trough) at each running spike time."""
    st = np.asarray(spike_times, dtype=float)
    if st.size == 0:
        return np.empty(0)
    # Keep spikes inside the theta time range and occurring during running
    st = st[(st >= ref_complex.index[0]) & (st <= ref_complex.index[-1])]
    if st.size == 0:
        return np.empty(0)
    running = speed_series.reindex(st, method="nearest").values >= SPIKE_SPEED_THRESH
    st = st[running]
    if st.size == 0:
        return np.empty(0)
    ph = np.angle(ref_complex.reindex(st, method="nearest").values)
    return np.degrees(ph) % 360

def circ_stats(phase_rad):
    """Mean angle, mean resultant length R, and Rayleigh p (Zar approx)."""
    n = len(phase_rad)
    if n == 0:
        return np.nan, 0.0, 1.0, 0
    C, S = np.cos(phase_rad).sum(), np.sin(phase_rad).sum()
    R = np.hypot(C, S) / n
    mean_angle = np.arctan2(S, C)
    Zr = n * R ** 2
    p = np.exp(-Zr) * (1 + (2 * Zr - Zr ** 2) / (4 * n))   # Rayleigh test (Zar)
    return mean_angle, R, p, n

# Per-unit mean phase + locking strength (R), and the pooled spike-phase distribution
per_unit = [] # (mean_angle_rad, R, n_spikes) for each unit that fires during running
pooled_phase_deg = []
for _, row in good_units.iterrows():
    pd_deg = spike_phases_deg(row["spike_times"])
    if pd_deg.size == 0:
        continue
    pooled_phase_deg.append(pd_deg)
    ma, R, _, n = circ_stats(np.radians(pd_deg))
    per_unit.append((ma, R, n))
pooled_phase_deg = np.concatenate(pooled_phase_deg) if pooled_phase_deg else np.empty(0)
pooled_mean_angle, pooled_R, pooled_p, pooled_n = circ_stats(np.radians(pooled_phase_deg))

print(f"{session_id}: {len(reference_electrodes)} reference electrodes {reference_electrodes}")
print(f"cross-channel coherence {coherence:.3f}")
print(f"decode distance vs theta phase: circular-linear r = {clc:.3f}, peaks at {peak_phase:.0f} deg "
      f"(0 = theta peak, 180 = trough)")
print(f"spike phase locking: {len(per_unit)} units, {pooled_n} running spikes | "
      f"pooled mean phase {np.degrees(pooled_mean_angle) % 360:.0f} deg, R = {pooled_R:.3f}, "
      f"Rayleigh p = {pooled_p:.1e}")

fig = plt.figure(figsize=(22, 7))
gs = fig.add_gridspec(2, 3, height_ratios=[1, 1.2])

# (top) reference electrodes on the probe geometry
ax_geo = fig.add_subplot(gs[0, :])
for elec, row in electrode_geometry.iterrows():
    if pd.isna(row["rel_x"]):
        continue
    is_ref = str(elec) in ref_set
    face = "crimson" if is_ref else ("white" if is_bad.get(elec, False) else "0.85")
    ax_geo.add_patch(Rectangle((row["rel_x"] - box_width / 2, row["rel_y"] - box_height / 2),
                     box_width, box_height, facecolor=face, edgecolor="0.6", linewidth=0.3))
unit_mask = electrode_geometry.index.astype(int).isin(set(unit_electrodes))
ax_geo.scatter(electrode_geometry.loc[unit_mask, "rel_x"], electrode_geometry.loc[unit_mask, "rel_y"],
               marker="*", s=50, c="white", edgecolors="black", linewidths=0.5, zorder=5, label="has unit")
ax_geo.scatter([], [], marker="s", color="crimson", label="reference electrode")
ax_geo.set_xlim(electrode_geometry["rel_x"].min() - box_width, electrode_geometry["rel_x"].max() + box_width)
ax_geo.set_ylim(electrode_geometry["rel_y"].min() - box_height, electrode_geometry["rel_y"].max() + box_height)
ax_geo.set_box_aspect(np.ptp(electrode_geometry["rel_y"]) / np.ptp(electrode_geometry["rel_x"]))
ax_geo.set_xlabel("rel_x (µm)")
ax_geo.set_ylabel("rel_y (µm)")
ax_geo.set_title(f"{session_id} theta reference electrodes")
ax_geo.legend(loc="upper left", fontsize=8)

# (bottom-left) each reference channel's mean phase (gray) + resultant (red): clustered = clean
ax_polar = fig.add_subplot(gs[1, 0], projection="polar")
for ph in channel_mean_phase:
    ax_polar.plot([0, ph], [0, 1], color="gray", alpha=0.5)
ax_polar.plot([0, np.angle(np.mean(np.exp(1j * channel_mean_phase)))], [0, 1], color="red", lw=2.5)
ax_polar.set_yticklabels([])
ax_polar.set_title(f"per-channel mean phase\n(coherence {coherence:.2f})", fontsize=9)

# (bottom-middle) spike -> theta phase: pooled rose + per-unit preferred-phase vectors.
# 0 = reference theta peak, 180 = trough. 
# CA1 pyramidal cells fire near the trough, so ideally we would see ~180 here
ax_spk = fig.add_subplot(gs[1, 1], projection="polar")
n_bins = 24
spk_edges = np.linspace(0, 2 * np.pi, n_bins + 1)
if pooled_n > 0:
    counts, _ = np.histogram(np.radians(pooled_phase_deg), bins=spk_edges)
    counts = counts / counts.max()                       # normalize so the rose fits in [0, 1]
    ax_spk.bar(spk_edges[:-1], counts, width=np.diff(spk_edges), align="edge",
               bottom=0.0, color="0.7", edgecolor="0.4", linewidth=0.3, zorder=1)
    # per-unit preferred phase: arrow length = locking strength R (longer = more locked)
    for ma, R, n in per_unit:
        ax_spk.plot([0, ma], [0, R], color="steelblue", alpha=0.5, lw=1, zorder=2)
    # pooled mean phase (thick red), length = pooled R
    ax_spk.plot([0, pooled_mean_angle], [0, pooled_R], color="red", lw=3, zorder=3)
ax_spk.set_theta_zero_location("E")
ax_spk.set_theta_direction(1)
ax_spk.set_xticks(np.radians([0, 90, 180, 270]))
ax_spk.set_xticklabels(["0\n(peak)", "90", "180\n(trough)", "270"], fontsize=8)
ax_spk.set_yticklabels([])
ax_spk.set_ylim(0, 1)
ax_spk.set_title(f"spike theta phase ({len(per_unit)} units, {pooled_n} spk)\n"
                 f"pooled {np.degrees(pooled_mean_angle) % 360:.0f}°, R={pooled_R:.2f}", fontsize=9)

# (bottom-right) decode distance vs averaged theta phase, two cycles
ax_sweep = fig.add_subplot(gs[1, 2])
x_two = np.concatenate([centers, centers + 360])
ax_sweep.errorbar(x_two, np.concatenate([mean_dist, mean_dist]),
                  yerr=np.concatenate([sem_dist, sem_dist]), marker="o", capsize=3)
ax_sweep.axhline(0, color="gray", ls="--", lw=1)
for pk in (0, 360, 720):
    ax_sweep.axvline(pk, color="green", ls=":", lw=1)
for tr in (180, 540):
    ax_sweep.axvline(tr, color="purple", ls=":", lw=1)
ax_sweep.set_xlabel("averaged theta phase (deg; 0 = peak, 180 = trough)")
ax_sweep.set_ylabel("mean decode distance (cm)\n(+ ahead / - behind)")
ax_sweep.set_title(f"decode distance vs theta phase (circ-lin r = {clc:.3f})")

plt.tight_layout()
save_fig(fig, f"{session_id}_theta_reference_manual_summary")
plt.show()
```

## Plot a timeseries of LFP / spikes / theta phase / sweeps

```python
# Load a saved theta reference electrode set, then run the timeseries cell below.
import numpy as np
import spyglass.common as sgc
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

from spyglass_hexmaze.hex_maze_decoding import (
    HexMazeDecodedPosition, HexMazeThetaV1, HexMazeThetaReference,
)
from spyglass_hexmaze.spikesorting_helpers import (
    fetch_good_units, fetch_electrode_geometry, get_raw_electrical_series,
)

nwb_file_name = "Lily20251217_.nwb"
REFERENCE_NAME = "theta_reference_1"

# Which run epoch to analyse. Frank lab sessions have 4-5 run epochs ("01_r1", "03_r2",
# "05_r3", ...); Berke lab sessions have a single "00_r1".
# This has to be pinned because theta and decode are keyed DIFFERENTLY: HexMazeThetaV1 keys
# on target_interval_list_name, HexMazeDecodedPosition keys on the epoch number. Restricting
# either one by nwb_file_name alone can silently land them on different epochs, in which case
# no spike falls inside the theta window and every downstream count comes out zero.
# Look the epoch number up from TaskEpoch so both restrictions describe the same epoch.
TARGET_INTERVAL = "05_r3"

epoch = (sgc.TaskEpoch & {"nwb_file_name": nwb_file_name,
                          "interval_list_name": TARGET_INTERVAL}).fetch1("epoch")
theta_key = {"nwb_file_name": nwb_file_name, "target_interval_list_name": TARGET_INTERVAL}
decode_key = {"nwb_file_name": nwb_file_name, "epoch": epoch}
print(f"{nwb_file_name}  interval {TARGET_INTERVAL} = epoch {epoch}")

RUN_SPEED = 10.0            # cm/s threshold defining "running"
WINDOW = 4.0                # seconds of running to display
START_TIME = None #1790     # set start tme of plot (or None to auto-pick)

# Fetch reference electrode set for this nwb
ref = HexMazeThetaReference & {**theta_key, "reference_name": REFERENCE_NAME}
reference_electrodes = list(ref.fetch1("electrode_ids"))
print(f"loaded '{REFERENCE_NAME}': {len(reference_electrodes)} electrodes {reference_electrodes}")

# Pick a single reference electrode to use for LFP etc for this plot
single_ref_electrode = reference_electrodes[0] 

session_id = nwb_file_name.replace("_.nwb", "")

# Save figures as PDFs, one folder per session (same convention as the rest of the notebook)
FIG_DIR = Path("figures") / session_id
FIG_DIR.mkdir(parents=True, exist_ok=True)
def save_fig(fig, name):
    """Save a figure as a PDF into FIG_DIR."""
    path = FIG_DIR / f"{name}.pdf"
    fig.savefig(path, bbox_inches="tight")
    print(f"saved {path}")

# Per-electrode analytic theta signal + decode dataframe (both time-indexed)
# fetch1_analytic_signal() returns the per-electrode complex signal as columns
# "electrode {e}_real" / "electrode {e}_imag", which the timeseries cell reads directly.
analytic = (HexMazeThetaV1 & theta_key).fetch1_analytic_signal()
decode_df = (HexMazeDecodedPosition & decode_key).fetch1_dataframe()

# Probe geometry (rel_x, rel_y) for each electrode, indexed by name (string)
electrode_geometry = fetch_electrode_geometry(nwb_file_name)

# Good units (with spike_times) for this session, from whichever pipeline this lab uses
good_units = fetch_good_units(nwb_file_name)
print(f"{len(good_units)} good units | analytic {analytic.shape} | decode_df {decode_df.shape}")


# Make big time-series plot! (from a single reference electrode):
# 1) raw trace    
# 2) theta-filtered LFP        
# 3) theta phase (0=peak,180=trough)
# 4) theta power        
# 5) spike raster (good units, by depth)   
# 6) MUA (binned spike rate)
# 7) decode distance    
# 8) decode spatial coverage               
# 9) speed
# Early vs late halves of each theta cycle are shaded as vertical bands, so
# we can line up when spikes / MUA / sweeps events fall within the theta cycle

# If no start time was set, plot a window of the longest continuous running bout
t = decode_df.index.values
running = decode_df["speed"].values >= RUN_SPEED
# Segment boundaries of contiguous running samples
change = np.diff(running.astype(int))
seg_starts = np.where(change == 1)[0] + 1
seg_ends = np.where(change == -1)[0] + 1
if running[0]:
    seg_starts = np.r_[0, seg_starts]
if running[-1]:
    seg_ends = np.r_[seg_ends, len(running)]
if START_TIME is not None:
    t0 = START_TIME
else:
    best = np.argmax(t[seg_ends - 1] - t[seg_starts])   # longest bout
    t0 = t[seg_starts[best]]
t1 = t0 + WINDOW
print(f"showing electrode {single_ref_electrode}, running window {t0:.1f}-{t1:.1f}s")

# Raw trace for this electrode from the nwb
import pynwb
from spyglass.common import Nwbfile
with pynwb.NWBHDF5IO(Nwbfile().get_abs_path(nwb_file_name), "r", load_namespaces=True) as io:
    nwbf = io.read()
    eseries = get_raw_electrical_series(nwbf)   # "ElectricalSeries" (Berke) or "e-series" (Frank)
    n_samp = eseries.data.shape[0]
    if eseries.timestamps is not None:
        t_first, t_last = float(eseries.timestamps[0]), float(eseries.timestamps[-1])
        fs = (n_samp - 1) / (t_last - t_first)
    else: # regularly-sampled series (rate + start)
        t_first, fs = float(eseries.starting_time), float(eseries.rate)
    i0 = max(0, int(round((t0 - t_first) * fs)))
    i1 = min(n_samp, int(round((t1 - t_first) * fs)))
    # data column -> electrode_id, then find our electrode's column
    etable = nwbf.electrodes.to_dataframe()
    col_electrode_ids = [int(etable.index[r]) for r in eseries.electrodes.data]
    col = col_electrode_ids.index(int(single_ref_electrode))
    raw_vals = np.asarray(eseries.data[i0:i1, col], dtype=float)
    raw_time = t_first + np.arange(i0, i1) / fs
raw_lfp = pd.Series(raw_vals, index=raw_time)

# Theta-filtered LFP, theta phase, and theta power from the analytic signal
# The analytic signal is complex: its real part is the theta-filtered LFP, its angle is the
# theta phase, and its squared magnitude is the instantaneous theta power
e = single_ref_electrode
real = np.asarray(analytic[f"electrode {e}_real"])
imag = np.asarray(analytic[f"electrode {e}_imag"])
theta_complex = pd.Series(real + 1j * imag, index=analytic.index)   # complex theta, time-indexed
theta_filt = pd.Series(theta_complex.values.real, index=theta_complex.index).loc[t0:t1]
theta_phase_deg = pd.Series(np.degrees(np.angle(theta_complex.values)) % 360,
                            index=theta_complex.index).loc[t0:t1]
theta_power = pd.Series(np.abs(theta_complex.values) ** 2,
                        index=theta_complex.index).loc[t0:t1]   # amplitude^2 (µV²)

# Break the 0<->360 sawtooth wrap so the phase line isn't joined by vertical streaks
phase_plot = theta_phase_deg.values.astype(float).copy()
phase_plot[1:][np.abs(np.diff(phase_plot)) > 180] = np.nan

# Early vs late theta-cycle halves, shaded on every panel for easy time alignment
# Theta phase runs 0=peak -> 180=trough -> 360=peak. We call the peak->trough half "early"
# (0-180 deg) and the trough->peak half "late" (180-360 deg)
EARLY_BAND = (0, 180)  # deg: "early" part of the theta cycle (peak -> trough)
LATE_BAND = (180, 360) # deg: "late" part of the theta cycle (trough -> peak)
EARLY_COLOR, LATE_COLOR = "tab:blue", "tab:orange"

# 0 = early half, 1 = late half, -1 = neither (only if the bands don't cover the whole cycle)
band = np.full(len(theta_phase_deg), -1, dtype=int)
_ph = theta_phase_deg.values
band[(_ph >= EARLY_BAND[0]) & (_ph < EARLY_BAND[1])] = 0
band[(_ph >= LATE_BAND[0]) & (_ph < LATE_BAND[1])] = 1
band_times = theta_phase_deg.index.values
# Times of each theta 0-phase (start of a new cycle / theta peak): the samples where the
# phase wraps from ~360 back to ~0. Marked as vertical lines on every panel so the theta
# cycle breaks line up with the raw/filtered/spike/decode traces.
cycle_breaks = band_times[1:][np.diff(_ph) < -180]

def shade_theta_bands(ax):
    """Shade early/late theta-cycle halves as vertical bands on one axis. Because every panel
    shares the same time axis, the bands line up across panels for easy visual comparison."""
    if len(band) == 0:
        return
    edges = np.where(np.diff(band) != 0)[0] + 1          # boundaries between contiguous runs
    for s, e_ in zip(np.r_[0, edges], np.r_[edges, len(band)]):
        b = band[s]
        if b < 0:                                        # skip samples in neither band
            continue
        ax.axvspan(band_times[s], band_times[e_ - 1],
                   color=(EARLY_COLOR if b == 0 else LATE_COLOR), alpha=0.10, lw=0, zorder=0)

# Decode distance + spatial coverage in the window
decode_dist = decode_df["decode_distance"].loc[t0:t1]
spatial_cov = decode_df["spatial_cov"].loc[t0:t1]

# Spike raster: every good unit, ordered by sort group (shank). We order by sort group
# rather than by electrode depth so the raster reads the same way for both labs -- v0
# sessions have no peak channel, so per-unit depth is only available for half our data.
# kind="stable" keeps each sort group's units in the order the sorter produced them.
gu = good_units.copy().sort_values("sort_group_id", kind="stable")
spike_rows = [np.asarray(st, float)[(np.asarray(st, float) >= t0) & (np.asarray(st, float) <= t1)]
              for st in gu["spike_times"]]

# MUA: population multiunit rate = all units' spikes binned over the window
MUA_BIN = 0.010                                  # s (10 ms bins)
all_spikes_win = np.concatenate(spike_rows) if spike_rows else np.empty(0)
mua_edges = np.arange(t0, t1 + MUA_BIN, MUA_BIN)
mua_centers = 0.5 * (mua_edges[:-1] + mua_edges[1:])
mua_counts, _ = np.histogram(all_spikes_win, bins=mua_edges)
# light Gaussian smoothing (sigma ~1.5 bins) so the rate is readable
_sig = 1.5
_k = np.exp(-0.5 * (np.arange(-4, 5) / _sig) ** 2); _k /= _k.sum()
mua_rate = np.convolve(mua_counts / MUA_BIN, _k, mode="same")   # spikes/s summed across units

# Speed in the window
speed_win = decode_df["speed"].loc[t0:t1]


# Stacked figure, shared time axis
fig, axes = plt.subplots(9, 1, figsize=(15, 18), sharex=True,
                         gridspec_kw={"height_ratios": [1, 1, 0.8, 0.8, 2, 1, 1, 1, 1]})

axes[0].plot(raw_lfp.index, raw_lfp.values, color="0.3", lw=0.6)
axes[0].set_ylabel("raw trace\n(µV)")
axes[0].set_title(f"{session_id}  electrode {single_ref_electrode}  ({t0:.1f}-{t1:.1f}s, running)")

axes[1].plot(theta_filt.index, theta_filt.values, color="tab:blue", lw=1.0)
axes[1].axhline(0, color="0.7", lw=0.5)
axes[1].set_ylabel("theta-filt\n(µV)")

axes[2].plot(theta_phase_deg.index, phase_plot, color="tab:purple", lw=1.0)
axes[2].axhline(0, color="green", ls=":", lw=1)     # theta peak
axes[2].axhline(180, color="purple", ls=":", lw=1)  # theta trough
axes[2].set_yticks([0, 180, 360])
axes[2].set_ylabel("theta phase\n(deg)")

axes[3].fill_between(theta_power.index, theta_power.values, color="tab:cyan", alpha=0.5, lw=0)
axes[3].plot(theta_power.index, theta_power.values, color="tab:cyan", lw=0.9)
axes[3].set_ylabel("theta power\n(µV²)")

axes[4].eventplot(spike_rows, colors="k", lineoffsets=np.arange(len(spike_rows)),
                  linelengths=0.8, linewidths=0.5)
axes[4].set_ylim(-1, len(spike_rows))
axes[4].set_ylabel("units\n(by depth)")

axes[5].fill_between(mua_centers, mua_rate, step="mid", color="tab:green", alpha=0.5)
axes[5].set_ylabel("MUA\n(spikes/s)")

axes[6].plot(decode_dist.index, decode_dist.values, color="tab:red", lw=1.0)
axes[6].axhline(0, color="0.7", lw=0.5)
axes[6].set_ylabel("decode dist\n(cm; +ahead)")

axes[7].plot(spatial_cov.index, spatial_cov.values, color="tab:orange", lw=1.0)
axes[7].set_ylabel("spatial cov")

axes[8].plot(speed_win.index, speed_win.values, color="tab:brown", lw=1.0)
axes[8].axhline(RUN_SPEED, color="0.7", ls=":", lw=1)   # running threshold
axes[8].set_ylabel("speed\n(cm/s)")
axes[8].set_xlabel("time (s)")
axes[8].set_xlim(t0, t1)

# Shade early/late theta-cycle halves on every panel so events can be lined up by phase
for ax in axes:
    shade_theta_bands(ax)
    for tb in cycle_breaks: # dashed line at each theta 0-phase (LFP peak)
        ax.axvline(tb, color="0.35", ls="--", lw=0.6, zorder=1)

# Legend for the phase bands (on the top panel only)
from matplotlib.patches import Patch
axes[0].legend(handles=[Patch(color=EARLY_COLOR, alpha=0.25, label="early theta (peak→trough)"),
                        Patch(color=LATE_COLOR, alpha=0.25, label="late theta (trough→peak)")],
               loc="upper right", fontsize=8, framealpha=0.9)

plt.tight_layout()
save_fig(fig, f"{session_id}_electrode{single_ref_electrode}_timeseries_{t0:.0f}s")
plt.show()
```

```python
# Scrollable version of the timeseries plot above: drag the "start" slider to move the
# window across the whole session, and hit the Play button to auto-scroll through time.
# This reuses everything already loaded in the cell above (run that first)
import ipywidgets as widgets
from IPython.display import display
from matplotlib.patches import Patch

# Open the NWB file ONCE and keep it open, so scrolling can read raw traces quickly
# (The cell above re-opened the file for a single window; here we open once and slice per
# window on demand. We also precompute the electrode-id -> data-column mapping a single time.)
_io = pynwb.NWBHDF5IO(Nwbfile().get_abs_path(nwb_file_name), "r", load_namespaces=True)
_nwbf = _io.read()
_eseries = get_raw_electrical_series(_nwbf)   # "ElectricalSeries" (Berke) or "e-series" (Frank)
_n_samp = _eseries.data.shape[0]
if _eseries.timestamps is not None:
    _t_first, _t_last = float(_eseries.timestamps[0]), float(_eseries.timestamps[-1])
    _fs = (_n_samp - 1) / (_t_last - _t_first)
else:  # regularly-sampled series (rate + start)
    _t_first, _fs = float(_eseries.starting_time), float(_eseries.rate)
_etable = _nwbf.electrodes.to_dataframe()
_col_electrode_ids = [int(_etable.index[r]) for r in _eseries.electrodes.data]  # data col -> electrode id

def read_raw_window(electrode, a, b):
    """Read the raw trace for one electrode over [a, b] seconds, as a time-indexed Series."""
    col = _col_electrode_ids.index(int(electrode))
    i0 = max(0, int(round((a - _t_first) * _fs)))
    i1 = min(_n_samp, int(round((b - _t_first) * _fs)))
    vals = np.asarray(_eseries.data[i0:i1, col], dtype=float)
    tt = _t_first + np.arange(i0, i1) / _fs
    return pd.Series(vals, index=tt)

# Precompute the unit ordering once (unchanged as we scroll).
# Same sort-group ordering as the cell above, so both rasters match for either lab.
_gu = good_units.copy().sort_values("sort_group_id", kind="stable")
_unit_spike_times = [np.asarray(st, float) for st in _gu["spike_times"]]  # one array per unit

# Full session time span, used to set the slider range
_tmin = float(decode_df.index.values[0])
_tmax = float(decode_df.index.values[-1])


def draw_window(t0, window, electrode):
    """Draw the full 9-panel stack for the window [t0, t0+window] on reference `electrode`.
    Same panels/logic as the static cell above, just wrapped so a slider can redraw it."""
    t1 = t0 + window

    # ---- raw trace (read straight from the open NWB) ----
    raw_lfp = read_raw_window(electrode, t0, t1)

    # ---- theta-filtered LFP, phase, power from the analytic signal ----
    # The analytic signal is complex: real part = theta-filtered LFP, angle = theta phase,
    # squared magnitude = instantaneous theta power.
    real = np.asarray(analytic[f"electrode {electrode}_real"])
    imag = np.asarray(analytic[f"electrode {electrode}_imag"])
    theta_complex = pd.Series(real + 1j * imag, index=analytic.index)
    theta_filt = pd.Series(theta_complex.values.real, index=theta_complex.index).loc[t0:t1]
    theta_phase_deg = pd.Series(np.degrees(np.angle(theta_complex.values)) % 360,
                                index=theta_complex.index).loc[t0:t1]
    theta_power = pd.Series(np.abs(theta_complex.values) ** 2,
                            index=theta_complex.index).loc[t0:t1]   # amplitude^2 (µV²)

    # Break the 0<->360 sawtooth wrap so the phase line isn't joined by vertical streaks
    phase_plot = theta_phase_deg.values.astype(float).copy()
    phase_plot[1:][np.abs(np.diff(phase_plot)) > 180] = np.nan

    # Early vs late theta-cycle halves (0-180 peak->trough = "early", 180-360 trough->peak = "late")
    EARLY_BAND, LATE_BAND = (0, 180), (180, 360)
    EARLY_COLOR, LATE_COLOR = "tab:blue", "tab:orange"
    band = np.full(len(theta_phase_deg), -1, dtype=int)  # 0=early, 1=late, -1=neither
    _ph = theta_phase_deg.values
    band[(_ph >= EARLY_BAND[0]) & (_ph < EARLY_BAND[1])] = 0
    band[(_ph >= LATE_BAND[0]) & (_ph < LATE_BAND[1])] = 1
    band_times = theta_phase_deg.index.values
    # Theta 0-phase crossings (start of each cycle / LFP peak): phase wraps ~360 -> ~0
    cycle_breaks = band_times[1:][np.diff(_ph) < -180] if len(_ph) else np.empty(0)

    def shade_theta_bands(ax):
        """Shade early/late theta-cycle halves as vertical bands so panels line up by phase."""
        if len(band) == 0:
            return
        edges = np.where(np.diff(band) != 0)[0] + 1
        for s, e_ in zip(np.r_[0, edges], np.r_[edges, len(band)]):
            b = band[s]
            if b < 0:
                continue
            ax.axvspan(band_times[s], band_times[e_ - 1],
                       color=(EARLY_COLOR if b == 0 else LATE_COLOR), alpha=0.10, lw=0, zorder=0)

    # ---- decode distance + spatial coverage in the window ----
    decode_dist = decode_df["decode_distance"].loc[t0:t1]
    spatial_cov = decode_df["spatial_cov"].loc[t0:t1]

    # ---- spike raster: each good unit, ordered by depth; clip spikes to the window ----
    spike_rows = [st[(st >= t0) & (st <= t1)] for st in _unit_spike_times]

    # ---- MUA: all units' spikes binned over the window, lightly smoothed ----
    MUA_BIN = 0.010  # s (10 ms bins)
    all_spikes_win = np.concatenate(spike_rows) if spike_rows else np.empty(0)
    mua_edges = np.arange(t0, t1 + MUA_BIN, MUA_BIN)
    mua_centers = 0.5 * (mua_edges[:-1] + mua_edges[1:])
    mua_counts, _ = np.histogram(all_spikes_win, bins=mua_edges)
    _sig = 1.5
    _k = np.exp(-0.5 * (np.arange(-4, 5) / _sig) ** 2); _k /= _k.sum()
    mua_rate = np.convolve(mua_counts / MUA_BIN, _k, mode="same")  # spikes/s summed across units

    # ---- speed in the window ----
    speed_win = decode_df["speed"].loc[t0:t1]

    # ---- stacked figure, shared time axis (same layout as the static cell) ----
    fig, axes = plt.subplots(9, 1, figsize=(15, 18), sharex=True,
                             gridspec_kw={"height_ratios": [1, 1, 0.8, 0.8, 2, 1, 1, 1, 1]})

    axes[0].plot(raw_lfp.index, raw_lfp.values, color="0.3", lw=0.6)
    axes[0].set_ylabel("raw trace\n(µV)")
    axes[0].set_title(f"{session_id}  electrode {electrode}  ({t0:.1f}-{t1:.1f}s)")

    axes[1].plot(theta_filt.index, theta_filt.values, color="tab:blue", lw=1.0)
    axes[1].axhline(0, color="0.7", lw=0.5)
    axes[1].set_ylabel("theta-filt\n(µV)")

    axes[2].plot(theta_phase_deg.index, phase_plot, color="tab:purple", lw=1.0)
    axes[2].axhline(0, color="green", ls=":", lw=1)     # theta peak
    axes[2].axhline(180, color="purple", ls=":", lw=1)  # theta trough
    axes[2].set_yticks([0, 180, 360])
    axes[2].set_ylabel("theta phase\n(deg)")

    axes[3].fill_between(theta_power.index, theta_power.values, color="tab:cyan", alpha=0.5, lw=0)
    axes[3].plot(theta_power.index, theta_power.values, color="tab:cyan", lw=0.9)
    axes[3].set_ylabel("theta power\n(µV²)")

    axes[4].eventplot(spike_rows, colors="k", lineoffsets=np.arange(len(spike_rows)),
                      linelengths=0.8, linewidths=0.5)
    axes[4].set_ylim(-1, len(spike_rows))
    axes[4].set_ylabel("units\n(by depth)")

    axes[5].fill_between(mua_centers, mua_rate, step="mid", color="tab:green", alpha=0.5)
    axes[5].set_ylabel("MUA\n(spikes/s)")

    axes[6].plot(decode_dist.index, decode_dist.values, color="tab:red", lw=1.0)
    axes[6].axhline(0, color="0.7", lw=0.5)
    axes[6].set_ylabel("decode dist\n(cm; +ahead)")

    axes[7].plot(spatial_cov.index, spatial_cov.values, color="tab:orange", lw=1.0)
    axes[7].set_ylabel("spatial cov")

    axes[8].plot(speed_win.index, speed_win.values, color="tab:brown", lw=1.0)
    axes[8].axhline(RUN_SPEED, color="0.7", ls=":", lw=1)  # running threshold
    axes[8].set_ylabel("speed\n(cm/s)")
    axes[8].set_xlabel("time (s)")
    axes[8].set_xlim(t0, t1)

    # Shade early/late theta halves + mark theta 0-phase on every panel
    for ax in axes:
        shade_theta_bands(ax)
        for tb in cycle_breaks:
            ax.axvline(tb, color="0.35", ls="--", lw=0.6, zorder=1)

    axes[0].legend(handles=[Patch(color=EARLY_COLOR, alpha=0.25, label="early theta (peak→trough)"),
                            Patch(color=LATE_COLOR, alpha=0.25, label="late theta (trough→peak)")],
                   loc="upper right", fontsize=8, framealpha=0.9)
    plt.tight_layout()
    plt.show()


# ---- Interactive controls ----
# start:     left edge of the window; drag or use the arrows to scroll across the whole session
# window:    how many seconds are shown at once
# electrode: which reference electrode to use for the raw/theta panels
# Play button auto-advances the start slider so you can watch time roll by.
_start = widgets.FloatSlider(value=_tmin, min=_tmin, max=max(_tmin, _tmax - WINDOW),
                             step=0.25, description="start (s)", continuous_update=False,
                             readout_format=".1f", layout=widgets.Layout(width="80%"))
_window = widgets.FloatSlider(value=WINDOW, min=1.0, max=20.0, step=0.5,
                              description="window (s)", continuous_update=False,
                              readout_format=".1f", layout=widgets.Layout(width="50%"))
_elec = widgets.Dropdown(options=[int(e) for e in reference_electrodes],
                         value=int(single_ref_electrode), description="electrode")
# Play advances `start` by its step; interval is ms between frames.
_play = widgets.Play(value=int(_tmin), min=int(_tmin), max=int(max(_tmin, _tmax - WINDOW)),
                     step=1, interval=700, description="scroll")
widgets.jslink((_play, "value"), (_start, "value"))

# Keep the start slider's max in sync with the chosen window width
def _sync_start_max(change):
    _start.max = max(_tmin, _tmax - _window.value)
    _play.max = int(max(_tmin, _tmax - _window.value))
_window.observe(_sync_start_max, names="value")

_out = widgets.interactive_output(draw_window, {"t0": _start, "window": _window, "electrode": _elec})
display(widgets.HBox([_play, _elec]), _start, _window, _out)
```
